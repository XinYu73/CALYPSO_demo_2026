```python
import os
import pandas as pd
from shutil import copy
```

```python
input_csv = "full.csv"
output_csv = "mp_synthesized_ABC3_with_O.csv"
os.makedirs('mp_synthesized_ABC3_with_O',exist_ok=True)
data = pd.read_csv(input_csv)


pattern = (
    r"^(?:"
    r"ABC3_[^:]+:[^-]+-[^-]+-O|"
    r"AB3C_[^:]+:[^-]+-O-[^-]+|"
    r"A3BC_[^:]+:O-[^-]+-[^-]+"
    r")$"
)

new_data = data[data["wyckoff"].str.fullmatch(pattern)]
new_data.to_csv(output_csv, index=False)

for x in new_data['mpID'].to_list():
    if os.path.exists(os.path.join('experimental',x+'.vasp')):
        copy(os.path.join('experimental',x+'.vasp'),os.path.join('mp_synthesized_ABC3_with_O',x+'.vasp'))
```
